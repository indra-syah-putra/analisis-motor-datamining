```python
data = data.drop(['Tipe_Penjual'], axis = 1)
data.info()
```

```python
from datetime import date
date.today().year
data['Umur_Motor']=date.today().year-data['Tahun']
data.head()
```

```python
data['Brand'] = data.Tipe_Motor.str.split().str.get(0)
data['Model'] = data.Tipe_Motor.str.split().str.get(1) + data.Tipe_Motor.str.split().str.get(2)
data[['Tipe_Motor', 'Brand', 'Model']]
```

```python
data.dropna(subset=['Model'], inplace=True)
print(data[['Tipe_Motor', 'Brand', 'Model']])
```

```python
import numpy as np

dataobject=data.select_dtypes(include=['object']).columns
datanumeric= data.select_dtypes(include=np.number).columns.tolist()
print("categorical variable:")
print(dataobject)
print("numerical variable:")
print(datanumeric)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
for col in datanumeric:
    print(col)
    print('skew :', round(data[col].skew(), 2))
    plt.figure(figsize= (15 , 4))
    plt.subplot(1, 2, 1)
    data[col].hist(grid=False)
    plt.ylabel('count')
    plt.subplot(1, 2, 2)
    sns.boxenplot(x=data[col])
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
object_stats = data.describe(include='object')
for col in object_stats:
    plt.figure(figsize=(8, 6))
    sns.countplot(y=data[col], orient='h', order=data[col].value_counts().index[:10])
    plt.title(f'Top 10 Counts of {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.show()
```

```python
fig, axes = plt.subplots(3, 2, figsize = (23, 23))
fig.suptitle('Bar plot for all categorical variables in the dataset')
sns.countplot(ax = axes[0, 0], x = 'Harga_penjualan', data = data, color= 'blue',
              order = data['Harga_penjualan'].head(15).value_counts().index);
sns.countplot(ax = axes[0, 1], x = 'Tipe_Pembeli', data = data, color= 'blue',
              order = data['Tipe_Pembeli'].value_counts().index);
sns.countplot(ax = axes[1, 0], x = 'harga_pameran', data = data, color= 'blue',
              order = data['harga_pameran'].head(20).value_counts().index);
sns.countplot(ax = axes[1, 1], x = 'Brand', data = data, color= 'blue',
              order = data['Brand'].value_counts().index);
sns.countplot(ax = axes[2, 0], x = 'Model', data = data, color= 'blue',
              order = data['Model'].head(10).value_counts().index);
sns.countplot(ax = axes[2, 1], x = 'Umur_Motor', data = data, color= 'blue',
              order = data['Umur_Motor'].value_counts().index);
axes[1][1].tick_params(labelrotation=45);
axes[2][0].tick_params(labelrotation=90);
axes[2][1].tick_params(labelrotation=90);
```
